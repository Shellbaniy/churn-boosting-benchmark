# 01 — EDA (только train)

Тонкая обёртка над функциями из `src/eda.py`. Тестовые строки не используются.

```python
import sys
from pathlib import Path

root = Path.cwd()
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from IPython.display import Image, display

from src.config import load_config
from src.eda import (
    class_balance,
    churn_by_category,
    load_train,
    load_train_raw,
    missing_totalcharges_table,
    plot_churn_rate,
    plot_class_imbalance,
    plot_missing_totalcharges,
    plot_tenure_by_class,
    print_conclusions,
    save_tables,
)
from src.utils import ensure_dir

cfg = load_config()
ensure_dir(cfg.figures_dir)
ensure_dir(cfg.tables_dir)

train = load_train(cfg)
train_raw = load_train_raw(cfg)
print(f'EDA только на train: {train.shape[0]} строк; тест не используется')

def show(path):
    display(Image(filename=str(path)))
```

## Дисбаланс классов

```python
balance = class_balance(train, cfg)
balance
show(plot_class_imbalance(balance, cfg))
```

## Пропуски TotalCharges и tenure

```python
miss = missing_totalcharges_table(train_raw, cfg)
miss
show(plot_missing_totalcharges(train_raw, cfg))
```

## Доля оттока по Contract

```python
contract = churn_by_category(train, cfg, 'Contract')
contract
show(plot_churn_rate(contract, cfg, 'Contract', 'тип контракта', 'eda_contract_churn.png'))
```

## Распределение tenure по классам

```python
show(plot_tenure_by_class(train, cfg))
```

## Доля оттока по InternetService

```python
internet = churn_by_category(train, cfg, 'InternetService')
internet
show(plot_churn_rate(internet, cfg, 'InternetService', 'тип интернет-услуги', 'eda_internetservice_churn.png'))
```

## Доля оттока по PaymentMethod

```python
payment = churn_by_category(train, cfg, 'PaymentMethod')
payment
show(plot_churn_rate(payment, cfg, 'PaymentMethod', 'способ оплаты', 'eda_paymentmethod_churn.png'))
```

## Выводы

```python
cat_tables = {'Contract': contract, 'InternetService': internet, 'PaymentMethod': payment}
save_tables(balance, miss, cat_tables, cfg)
print_conclusions(balance, train_raw, cat_tables, cfg)
```
